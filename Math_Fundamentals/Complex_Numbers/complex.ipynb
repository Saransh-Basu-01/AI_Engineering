Complex class

```python
import math 

class Complex:
    def __init__(self,real,imag=0.0,op=None):
        self.real=real
        self.imag=imag
        self.op=op
        
    def __add__(self,other):
        return Complex(self.real + other.real,self.imag+other.imag,op="+")
    
    def __sub__(self, other):
        return Complex(self.real - other.real,
                       self.imag - other.imag, op="-")
    
    def __mul__(self, other):
        r = self.real * other.real - self.imag * other.imag
        i = self.real * other.imag + self.imag * other.real
        return Complex(r, i,op="*")
    
    def __truediv__(self, other):
        denom = other.real ** 2 + other.imag ** 2
        r = (self.real * other.real + self.imag * other.imag) / denom
        i = (self.imag * other.real - self.real * other.imag) / denom
        return Complex(r, i,op="/")

    def magnitude(self):
        return Complex(math.sqrt(self.real**2 + self.imag**2), 0, op="magnitude")

    def phase(self):
        return Complex(math.atan2(self.imag, self.real), 0, op="phase")

    def conjugate(self):
        return Complex(self.real, -self.imag, op="conjugate")
    
    @staticmethod
    def to_polar(z):
        return Complex(z.magnitude().real, z.phase().real, op="to_polar")

    @staticmethod
    def from_polar(r, theta):
        return Complex(r * math.cos(theta), r * math.sin(theta), op="from_polar")
    
    @staticmethod
    def euler(theta):
        return Complex(math.cos(theta), math.sin(theta), op="euler")
    
    
    @staticmethod
    def dft(signal):
        N = len(signal)
        result = []
        for k in range(N):
            total = Complex(0, 0)
            for n in range(N):
                angle = -2 * math.pi * k * n / N
                total = total + Complex(signal[n], 0) * Complex.euler(angle)
            result.append(total)
        return result

    @staticmethod
    def idft(spectrum):
        N = len(spectrum)
        result = []
        for n in range(N):
            total = Complex(0, 0)
            for k in range(N):
                angle = 2 * math.pi * k * n / N
                total = total + spectrum[k] * Complex.euler(angle)
            result.append(Complex(total.real / N, total.imag / N))
        return result

    @staticmethod
    def roots_of_unity(N):
        return [Complex.euler(2 * math.pi * k / N) for k in range(N)]
    
    def __repr__(self):
        if self.op is None:
            return f"Complex({self.real}, {self.imag})"
        elif self.op == "+":
            return f"add → {self.real} + {self.imag}i"
        elif self.op == "-":
            return f"sub → {self.real} + {self.imag}i"
        elif self.op == "*":
            return f"mul → {self.real} + {self.imag}i"
        elif self.op == "/":
            return f"div → {self.real} + {self.imag}i"
        elif self.op == "magnitude":
            return f"magnitude → {self.real}"
        elif self.op == "phase":
            return f"phase → {self.real}"
        elif self.op == "conjugate":
            return f"conjugate → {self.real} + {self.imag}i"
        elif self.op == "to_polar":
            return f"to_polar → (r={self.real}, θ={self.imag})"
        elif self.op == "from_polar":
            return f"from_polar → {self.real} + {self.imag}i"
        elif self.op == "euler":
            return f"euler → {self.real} + {self.imag}i"
        else:
            return f"unknown → {self.real}, {self.imag}"
    
    
    
```

```python
import math

a = Complex(3, 4)
b = Complex(1, 2)

# ── Arithmetic ────────────────────────────────
print(a + b)          # → add → 4 + 6i
print(a - b)          # → sub → 2 + 2i
print(a * b)          # → mul → -5 + 10i
print(a / b)          # → div → 2.2 - 0.4i

# ── Unary operations ──────────────────────────
print(a.magnitude())  # → magnitude → 5.0
print(a.phase())      # → phase → 0.9272952180016122
print(a.conjugate())  # → conjugate → 3 + -4i

# ── Plain object (no op) ──────────────────────
print(a)              # → Complex(3, 4)
print(b)              # → Complex(1, 2)

# ── Polar helpers ─────────────────────────────
print(a.to_polar())                         # → to_polar → (r=5.0, θ=0.9272952180016122)
print(Complex.from_polar(5, math.pi / 4))   # → from_polar → 3.5355... + 3.5355...i
print(Complex.euler(math.pi))               # → euler → -1.0 + 1.2246e-16i

# ── Other test pairs ──────────────────────────
c = Complex(0, 1)     # pure imaginary (i)
d = Complex(0, -1)    # -i
e = Complex(5, 0)     # pure real
f = Complex(0, 0)     # origin

print(c)              # → Complex(0, 1)
print(c.magnitude())  # → magnitude → 1.0
print(c.phase())      # → phase → 1.5707963267948966  (π/2)
print(c.conjugate())  # → conjugate → 0 + -1i

print(d.magnitude())  # → magnitude → 1.0
print(d.phase())      # → phase → -1.5707963267948966  (-π/2)

print(e.magnitude())  # → magnitude → 5.0
print(e.phase())      # → phase → 0.0
print(e.conjugate())  # → conjugate → 5 + 0i

print(f.magnitude())  # → magnitude → 0.0
print(f.phase())      # → phase → 0.0
print(f.conjugate())  # → conjugate → 0 + 0i

# ── Mixed signs ───────────────────────────────
g = Complex(-3, 4)    # quadrant II
h = Complex(-3, -4)   # quadrant III
i = Complex(3, -4)    # quadrant IV

print(g.magnitude())  # → magnitude → 5.0
print(g.phase())      # → phase → 2.214297435588181   (π - atan(4/3))
print(g.conjugate())  # → conjugate → -3 + -4i

print(h.magnitude())  # → magnitude → 5.0
print(h.phase())      # → phase → -2.214297435588181

print(i.magnitude())  # → magnitude → 5.0
print(i.phase())      # → phase → -0.9272952180016122
```
