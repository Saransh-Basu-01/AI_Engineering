# Q1): Implement Vector.angle_between(other) that returns the angle in degrees between two vectors

```python
import numpy as np
class Vector:
    def __init__(self,components):
        self.components=list(components)
        self.ndim=len(components)
        print(self.components)
    
    def angle_between(self,other):
        dot_product=np.dot(self.components,other.components)
        norm_v1=np.linalg.norm(self.components)
        norm_v2=np.linalg.norm(other.components)
        cos_angle=dot_product/(norm_v1*norm_v2)
        cos_angle_clipped = np.clip(cos_angle, -1.0, 1.0)
        angle_radians = np.arccos(cos_angle_clipped)
        angle_degrees = np.degrees(angle_radians)
        return angle_degrees
        
a = Vector([1, 2, 3])
b = Vector([4, 5, 6])
angle=a.angle_between(b)
print(f"The angle in degrees is {angle}")
```

class Vector:
    def __init__(self, components):
        # Convert input to list to ensure it's mutable and works with various iterables
        self.components = list(components)
        # Store the dimension of the vector
        self.ndim = len(components)
        # Print the components (useful for debugging)
        print(self.components)
    
    def angle_between(self, other):
        # Calculate dot product using numpy for efficiency
        # This gives us the sum of element-wise products: v1·v2 = Σ(v1i * v2i)
        dot_product = np.dot(self.components, other.components)
        
        # Calculate Euclidean norms (magnitudes) of both vectors
        # ||v|| = √(Σ(vi²)), used for normalization
        norm_v1 = np.linalg.norm(self.components)
        norm_v2 = np.linalg.norm(other.components)
        
        # Calculate cosine of the angle using the formula:
        # cos(θ) = (v1·v2) / (||v1|| * ||v2||)
        # This gives us a value between -1 and 1 in theory
        cos_angle = dot_product / (norm_v1 * norm_v2)
        
        # CLIP: Why use clip?
        # Due to floating-point precision errors, cos_angle might be slightly 
        # outside the theoretical range [-1, 1] (e.g., 1.0000000000000002 or -1.0000000000000002)
        # This would cause np.arccos() to fail (return nan) since it only accepts values in [-1, 1]
        # Clip ensures we stay within this valid range
        cos_angle_clipped = np.clip(cos_angle, -1.0, 1.0)
        
        # Calculate the angle in radians using inverse cosine (arccos)
        # arccos(cos(θ)) = θ, giving us the angle in radians
        angle_radians = np.arccos(cos_angle_clipped)
        
        # Convert radians to degrees for a more intuitive representation
        # Since most people think in degrees (0° to 180°), not radians (0 to π)
        angle_degrees = np.degrees(angle_radians)
        
        return angle_degrees

# Q2):Create a 2D scaling matrix that doubles the x-coordinate and triples the y-coordinate, then apply it to the vector [1, 1]

```python
import numpy as np

# Create the scaling matrix
# For 2D scaling, the matrix is:
# [sx  0]
# [0   sy]
# where sx is the scaling factor for x-axis and sy for y-axis
scaling_matrix = np.array([
    [2, 0],  # Doubles x-coordinate
    [0, 3]   # Triples y-coordinate
])

# Define the original vector
vector = np.array([1, 1])

# Apply the scaling transformation
# Matrix multiplication: result = scaling_matrix × vector
# For a 2x2 matrix and 2D vector, this gives:
# [2  0] × [1] = [2*1 + 0*1] = [2]
# [0  3]   [1]   [0*1 + 3*1]   [3]
scaled_vector = np.dot(scaling_matrix, vector)

# Alternative using @ operator (Python 3.5+)
# scaled_vector = scaling_matrix @ vector

print(f"Original vector: {vector}")
print(f"Scaling matrix:\n{scaling_matrix}")
print(f"Scaled vector: {scaled_vector}")
```

# Q3):Given 5 random word-like vectors (dimension 50), find the two most similar using cosine similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
np.random.seed(42)
word_labels = ['apple', 'banana', 'cat', 'dog', 'elephant']
print("Generated word vectors (showing first 5 dimensions):")
word_vectors = np.random.randn(5, 50)
for i, label in enumerate(word_labels):
    print(f"{label}: {word_vectors[i][:5]}...")  # Show only first 5 dims for brevity
print()
```

##  Method 1: Using sklearn's cosine_similarity

```python

similarity_matrix=cosine_similarity(word_vectors)
similarities=[]
for i in range(len(word_vectors)):
    for j in range(i+1,len(word_vectors)):
        similarities.append((i,j,similarity_matrix[i][j]))
similarities.sort(key=lambda x: x[2], reverse=True)
most_similar_pair = similarities[0]
i, j, score = most_similar_pair
print("="*50)
print("METHOD 1: Using sklearn's cosine_similarity")
print("="*50)
print(f"Most similar pair: '{word_labels[i]}' and '{word_labels[j]}'")
print(f"Cosine similarity score: {score:.4f}")
print()
print(len(similarities))
```

##  Method 2: Manual implementation using numpy

```python
def cosine_similarity_manual(v1, v2):
    """Calculate cosine similarity between two vectors"""
    dot_product = np.dot(v1, v2)
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)
    return dot_product / (norm_v1 * norm_v2)

# Calculate all pairwise similarities manually
manual_similarities = []
for i in range(len(word_vectors)):
    for j in range(i+1, len(word_vectors)):
        sim = cosine_similarity_manual(word_vectors[i], word_vectors[j])
        manual_similarities.append((i, j, sim))

manual_similarities.sort(key=lambda x: x[2], reverse=True)
i_manual, j_manual, score_manual = manual_similarities[0]

print("="*50)
print("METHOD 2: Manual implementation with numpy")
print("="*50)
print(f"Most similar pair: '{word_labels[i_manual]}' and '{word_labels[j_manual]}'")
print(f"Cosine similarity score: {score_manual:.4f}")
print()
```

# Q4):Verify that the Gram-Schmidt output is truly orthonormal: check that every pair has dot product 0 and every vector has magnitude 1


```python
import math

class Vector:
    def __init__(self, components):
        self.components = list(components)

    def dot(self, other):
        return sum(a * b for a, b in zip(self.components, other.components))

    def magnitude(self):
        return math.sqrt(self.dot(self))

    def normalize(self):
        m = self.magnitude()
        return Vector([x / m for x in self.components])

    def __sub__(self, other):
        return Vector([a - b for a, b in zip(self.components, other.components)])

    def __repr__(self):
        return f"Vector({[round(x, 10) for x in self.components]})"


def project(a, b):
    scalar = a.dot(b) / b.dot(b)
    return Vector([scalar * x for x in b.components])


def gram_schmidt(vectors):
    orthonormal = []
    for v in vectors:
        w = v
        for u in orthonormal:
            proj = project(w, u)
            w = w - proj
        if w.magnitude() < 1e-10:
            continue
        orthonormal.append(w.normalize())
    return orthonormal


def verify(orthonormal, tol=1e-10):
    print(f"Got {len(orthonormal)} orthonormal vectors\n")

    ok = True
    # 1) every vector has magnitude 1
    print("--- Magnitudes (should be 1) ---")
    for i, u in enumerate(orthonormal):
        m = u.magnitude()
        good = abs(m - 1) < tol
        ok &= good
        print(f"||u_{i+1}|| = {m:.12f}  {'✓' if good else '✗'}")

    # 2) every pair has dot product 0
    print("\n--- Pairwise dot products (should be 0) ---")
    for i in range(len(orthonormal)):
        for j in range(i + 1, len(orthonormal)):
            d = orthonormal[i].dot(orthonormal[j])
            good = abs(d) < tol
            ok &= good
            print(f"<u_{i+1}, u_{j+1}> = {d: .3e}  {'✓' if good else '✗'}")

    print("\n" + ("ALL CHECKS PASSED ✓" if ok else "FAILED ✗"))
    return ok


# --- Test on your example ---
vectors = [
    Vector([1, 1, 0]),
    Vector([1, 0, 1]),
    Vector([0, 1, 1]),
]

orthonormal = gram_schmidt(vectors)
for u in orthonormal:
    print(u)
print()
verify(orthonormal)
```
