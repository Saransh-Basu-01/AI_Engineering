Complex class

```python
import math 

class Complex:
    def __init__(self,real,imag=0.0,op=None):
        self.real=real
        self.imag=imag
        self.op=op
        
    def __add__(self,other):
        return Complex(self.real + other.real,self.imag+other.imag,op="+")
    
    def __sub__(self, other):
        return Complex(self.real - other.real,
                       self.imag - other.imag, op="-")
    
    def __mul__(self, other):
        r = self.real * other.real - self.imag * other.imag
        i = self.real * other.imag + self.imag * other.real
        return Complex(r, i,op="*")
    
    def __truediv__(self, other):
        denom = other.real ** 2 + other.imag ** 2
        r = (self.real * other.real + self.imag * other.imag) / denom
        i = (self.imag * other.real - self.real * other.imag) / denom
        return Complex(r, i,op="/")

    def magnitude(self):
        return Complex(math.sqrt(self.real**2 + self.imag**2), 0, op="magnitude")

    def phase(self):
        return Complex(math.atan2(self.imag, self.real), 0, op="phase")

    def conjugate(self):
        return Complex(self.real, -self.imag, op="conjugate")
    
    def __repr__(self):
        if self.op is None:
            return f"Complex({self.real}, {self.imag})"
        elif self.op == "+":
            return f"add → {self.real} + {self.imag}i"
        elif self.op == "-":
            return f"sub → {self.real} + {self.imag}i"
        elif self.op == "*":
            return f"mul → {self.real} + {self.imag}i"
        elif self.op == "/":
            return f"div → {self.real} + {self.imag}i"
        elif self.op == "magnitude":
            return f"magnitude → {self.real}"
        elif self.op == "phase":
            return f"phase → {self.real}"
        elif self.op == "conjugate":
            return f"conjugate → {self.real} + {self.imag}i"
        else:
            return f"unknown → {self.real}, {self.imag}"
    
    
    
```

```python
a = Complex(3, 4)
b = Complex(1, 2)

print(a + b)          # → add → 4 + 6i
print(a - b)          # → sub → 2 + 2i
print(a * b)          # → mul → -5 + 10i
print(a / b)          # → div → 2.2 - 0.4i
print(a.magnitude())  # → magnitude → 5.0
print(a.phase())      # → phase → 0.9272952180016122
print(a.conjugate())  # → conjugate → 3 + -4i
print(a)              # → Complex(3, 4)
```
