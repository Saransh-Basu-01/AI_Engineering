 Hinge loss and gradient

```python
import numpy as np
```

```python
def hinge_loss(X,y,w,b):
    n=X.shape[0]
    total_loss=0.0
    margin=y*(np.dot(w,X.T)+b)
    total_loss = np.sum(np.maximum(0.0, 1.0 - margin))
    return total_loss/n
```

```python
# Simple test case
X = np.array([[1.0, 2.0],   # sample 1
     [2.0, 1.0],   # sample 2
     [0.5, 0.5]] )  # sample 3

y = np.array([[1, -1, 1]]  )   # labels (+1 / -1)
w = np.array([[1.0, 1.0]] )    # weight vector
b = np.array([[0.0]])            # bias

print(hinge_loss(X, y, w, b))
```
