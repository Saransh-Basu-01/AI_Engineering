```python
import torch
import torch.nn as nn
import torch.optim as optim

# Model
model = torch.nn.Linear(784, 10)

# Optimizers
sgd = torch.optim.SGD(model.parameters(), lr=0.01, momentum=0.9)
adam = torch.optim.Adam(model.parameters(), lr=0.001)
adamw = torch.optim.AdamW(model.parameters(), lr=0.001, weight_decay=0.01)

# Scheduler (attached to Adam)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(adam, T_max=100)

# Dummy data
batch_size = 32
inputs = torch.randn(batch_size, 784)
targets = torch.randint(0, 10, (batch_size,))

# Loss function
criterion = nn.CrossEntropyLoss()

# ---- Training example using Adam + CosineAnnealingLR ----
num_epochs = 100
for epoch in range(num_epochs):
    # Forward
    outputs = model(inputs)
    loss = criterion(outputs, targets)

    # Backward
    adam.zero_grad()
    loss.backward()
    adam.step()

    # Scheduler step (after optimizer step)
    scheduler.step()

    current_lr = scheduler.get_last_lr()[0]
    print(f"Epoch [{epoch+1}/{num_epochs}]  Loss: {loss.item():.4f}  LR: {current_lr:.6f}")

# ---- Optional: If you want to use SGD or AdamW instead ----
def train_with_optimizer(optimizer, scheduler=None, epochs=10):
    """Generic training loop for any optimizer."""
    for epoch in range(epochs):
        outputs = model(inputs)
        loss = criterion(outputs, targets)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        if scheduler is not None:
            scheduler.step()
            lr = scheduler.get_last_lr()[0]
        else:
            lr = optimizer.param_groups[0]['lr']

        print(f"[{optimizer.__class__.__name__}] Epoch [{epoch+1}/{epochs}] "
              f"Loss: {loss.item():.4f}  LR: {lr:.6f}")

# Example: Train with SGD
# train_with_optimizer(sgd, epochs=10)

# Example: Train with AdamW
# train_with_optimizer(adamw, epochs=10)
```
