 Hinge loss and gradient

```python
import numpy as np
```

```python
def hinge_loss(X,y,w,b):
    n=X.shape[0]
    total_loss=0.0
    margin=y*(np.dot(w,X.T)+b)
    total_loss = np.sum(np.maximum(0.0, 1.0 - margin))
    return total_loss/n
```

```python
# Simple test case
X = np.array([[1.0, 2.0],   # sample 1
     [2.0, 1.0],   # sample 2
     [0.5, 0.5]] )  # sample 3

y = np.array([[1, -1, 1]]  )   # labels (+1 / -1)
w = np.array([[1.0, 1.0]] )    # weight vector
b = np.array([[0.0]])            # bias

print(hinge_loss(X, y, w, b))
```

```python
import numpy as np

class LinearSVM:
    def __init__(self, lr=0.01, lambda_param=0.01, n_epochs=1000):
        self.lr = lr
        self.lambda_param = lambda_param
        self.n_epochs = n_epochs
        self.w = None
        self.b = 0.0

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.w = np.zeros(n_features)
        self.b = 0.0

        for epoch in range(self.n_epochs):
            for i in range(n_samples):
                margin = y[i] * (np.dot(X[i], self.w) + self.b)
                if margin >= 1:
                    self.w -= self.lr * self.lambda_param * self.w
                    # b unchanged (gradient is 0)
                else:
                    self.w -= self.lr * (self.lambda_param * self.w - y[i] * X[i])
                    self.b -= self.lr * (-y[i])

    def predict(self, X):
        scores = np.dot(X, self.w) + self.b
        return np.where(scores >= 0, 1, -1)
```

```python
X = np.array([[2.0, 2.0],
     [2.5, 2.5],     # class +1
     [1.5, 2.0],
     [0.0, 0.0],
     [0.5, 0.0],     # class -1
     [-0.5, -0.5]])
y = np.array([1, 1, 1, -1, -1, -1]) 

model = LinearSVM(lr=0.01, lambda_param=0.01, n_epochs=1000)
model.fit(X, y)

print("w =", model.w)
print("b =", model.b)
print("predictions:", model.predict(X))
print("expected   :", y)
```
